```python
grupos_categorias = {
    "Sin clasificacion tematica": ["Default", "Add a comment"],
    "Ficcion general y humor": ["Fiction", "Contemporary", "Novels", "Short Stories", "Adult Fiction", "Humor"],
    "Fantasia, ciencia ficcion y terror": ["Fantasy", "Science Fiction", "Paranormal", "Horror"],
    "Misterio, crimen y suspenso": ["Mystery", "Thriller", "Crime", "Suspense"],
    "Romance y relaciones": ["Romance", "Erotica", "Womens Fiction", "New Adult"],
    "Ficcion historica": ["Historical Fiction", "Historical"],
    "Literatura infantil y juvenil": ["Childrens", "Young Adult"],
    "Comics y narrativa grafica": ["Sequential Art"],
    "Poesia y clasicos": ["Poetry", "Classics"],
    "No ficcion general": ["Nonfiction"],
    "Historia, sociedad y biografias": ["History", "Biography", "Autobiography", "Politics", "Cultural"],
    "Filosofia, religion y espiritualidad": ["Philosophy", "Religion", "Spirituality", "Christian", "Christian Fiction"],
    "Arte y musica": ["Art", "Music"],
    "Desarrollo personal, educacion y negocios": ["Self Help", "Psychology", "Parenting", "Academic", "Business"],
    "Ciencia, salud y deporte": ["Science", "Health", "Sports and Games"],
    "Viajes y gastronomia": ["Travel", "Food and Drink"],
}
mapa_grupos={}
for grupo, categorias in grupos_categorias.items():
    for categoria in categorias:
        mapa_grupos[categoria]=grupo
df["Categoria_grupo"]=df["Categoria"].map(mapa_grupos)
```

```python
df=df.drop(columns=["Categoria"])
```
